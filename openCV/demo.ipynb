```python
def rescaleFrame(frame, scale=0.5):
    width = int(frame.shape[1] * scale)
    height = int(frame.shape[0] * scale)
    dimensions = (width, height)

    return cv.resize(frame, dimensions, interpolation=cv.INTER_AREA)
    
```

```python
img = cv.imread("Photos/cat1.jpg")

# cv.namedWindow("Cat", cv.WINDOW_NORMAL)

# cv.resizeWindow("Cat", 800, 600)

cv.imshow('Cat', img)
resized_img = rescaleFrame(img)
cv.imshow('Cat_resize', resized_img)

cv.waitKey(0)
```

#### Canny detects strong changes in intensity and produces edges.

#### Blurring makes the image smoother.

#### Grayscale is commonly used before operations such as:
Canny edge detection, 
Thresholding, 
Face detection

#### Dilation makes white regions/edges thicker. 

#### Erosion does almost the opposite of dilation. 

```python
import cv2 as cv

# Read image
img = cv.imread("Photos/cat1.jpg")

if img is None:
    print("Image not found!")
    exit()

# ==============================
# 1. Grayscale
# ==============================

gray = cv.cvtColor(img, cv.COLOR_BGR2GRAY)


# ==============================
# 2. Blur
# ==============================

blur = cv.GaussianBlur(
    img,
    (7, 7),
    0
)


# ==============================
# 3. Edge Detection
# ==============================

edges = cv.Canny(
    gray,
    100,
    200
)


# ==============================
# 4. Dilation
# ==============================

dilated = cv.dilate(
    edges,
    (7, 7),
    iterations=3
)


# ==============================
# 5. Erosion
# ==============================

eroded = cv.erode(
    dilated,
    (7, 7),
    iterations=3
)


# ==============================
# 6. Resize
# ==============================

resized = cv.resize(
    img,
    (500, 400)
)


# ==============================
# 7. Crop
# ==============================

cropped = img[
    100:400,
    150:500
]


# ==============================
# Display
# ==============================

cv.imshow("Original", img)
cv.imshow("Grayscale", gray)
cv.imshow("Blur", blur)
cv.imshow("Edges", edges)
cv.imshow("Dilated", dilated)
cv.imshow("Eroded", eroded)
cv.imshow("Resized", resized)
cv.imshow("Cropped", cropped)

cv.waitKey(0)
cv.destroyAllWindows()
```

# Image changes

```python
import numpy as np
import cv2 as cv

def translate(img, x, y):
    transMat = np.float32([[1,0,x],[0,1,y]])
    dimensions = (img.shape[1], img.shape[0])
    return cv.warpAffine(img, transMat, dimensions)

def show_image(window_name, image, width=800, height=600):

    cv.namedWindow(window_name, cv.WINDOW_NORMAL)
    cv.resizeWindow(window_name, width, height)
    cv.imshow(window_name, image)

def rotate(img, angle, rotPoint=None):
    (height, width) = img.shape[:2]

    if rotPoint is None:
        rotPoint = (width/2, height/2)

    rotMat = cv.getRotationMatrix2D(rotPoint, angle, 1.0)
    dimensions = (width, height)

    return cv.warpAffine(img, rotMat, dimensions)
```

```python
import cv2 as cv

img = cv.imread("Photos/dog2.jpg")

if img is None:
    print("Image not found!")
else:
    cv.namedWindow("Cat", cv.WINDOW_NORMAL)
    cv.resizeWindow("Cat", 800, 600)

    show_image("Cat", img)

    translated = translate(img, -100, 100)

    show_image('translated', translated)
    
    cv.waitKey(0)
    cv.destroyAllWindows()
```
