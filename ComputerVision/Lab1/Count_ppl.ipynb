```python
import cv2

# Load Haar Cascade face detector
face_cascade = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_default.xml"
)

print("1. Image")
print("2. Video File")
print("3. Webcam")

choice = input("Enter your choice: ")


# Function to resize only for display
def resize_for_display(frame, max_width=800, max_height=550):

    height, width = frame.shape[:2]

    scale = min(
        max_width / width,
        max_height / height,
        1
    )

    new_width = int(width * scale)
    new_height = int(height * scale)

    return cv2.resize(frame, (new_width, new_height))


if choice == "1":

    image_path = input("Enter image path: ")

    image = cv2.imread(image_path)

    if image is None:
        print("Image not found!")
        exit()

    # Convert to grayscale
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

    # Improve contrast
    gray = cv2.equalizeHist(gray)

    # Detect faces
    faces = face_cascade.detectMultiScale(
        gray,
        scaleFactor=1.05,
        minNeighbors=8,
        minSize=(40, 40)
    )

    print("Number of faces:", len(faces))

    # Draw rectangles
    for (x, y, w, h) in faces:

        cv2.rectangle(
            image,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )

    # Display face count
    cv2.putText(
        image,
        "Faces: " + str(len(faces)),
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0, 255, 0),
        2
    )

    # Resize ONLY for displaying
    display_image = resize_for_display(image)

    cv2.namedWindow(
        "Face Detection",
        cv2.WINDOW_NORMAL
    )

    cv2.imshow(
        "Face Detection",
        display_image
    )

    cv2.waitKey(0)
    cv2.destroyAllWindows()



elif choice == "2" or choice == "3":

    if choice == "2":

        video_path = input("Enter video path: ")

        cap = cv2.VideoCapture(video_path)

    else:

        cap = cv2.VideoCapture(0)

    if not cap.isOpened():
        print("Unable to open video!")
        exit()

    cv2.namedWindow(
        "Face Detection",
        cv2.WINDOW_NORMAL
    )

    while True:

        ret, frame = cap.read()

        if not ret:
            break

        # Convert to grayscale
        gray = cv2.cvtColor(
            frame,
            cv2.COLOR_BGR2GRAY
        )

        # Improve contrast
        gray = cv2.equalizeHist(gray)

        # Detect faces
        faces = face_cascade.detectMultiScale(
            gray,
            scaleFactor=1.05,
            minNeighbors=8,
            minSize=(40, 40)
        )

        # Draw rectangles
        for (x, y, w, h) in faces:

            cv2.rectangle(
                frame,
                (x, y),
                (x + w, y + h),
                (0, 255, 0),
                2
            )

        # Display face count
        cv2.putText(
            frame,
            "Faces: " + str(len(faces)),
            (20, 40),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0, 255, 0),
            2
        )

        # Resize ONLY for display
        display_frame = resize_for_display(frame)

        cv2.imshow(
            "Face Detection",
            display_frame
        )

        # ESC to exit
        if cv2.waitKey(1) == 27:
            break

    cap.release()
    cv2.destroyAllWindows()


else:
    print("Invalid choice!")
```
