```python
import cv2

# Function to resize image only for display
def resize_for_display(image, max_width=700, max_height=500):

    height, width = image.shape[:2]

    scale = min(
        max_width / width,
        max_height / height,
        1
    )

    new_width = int(width * scale)
    new_height = int(height * scale)

    return cv2.resize(image, (new_width, new_height))


# Get image from user
image_path = input("Enter image path: ")

image = cv2.imread(image_path)

if image is None:
    print("Image not found!")
    exit()

print("\nChoose an operation:")
print("1. Resize")
print("2. Rotate")
print("3. Blur")

choice = input("Enter your choice: ")


# ================= RESIZE =================

if choice == "1":

    width = int(input("Enter new width: "))
    height = int(input("Enter new height: "))

    resized = cv2.resize(image, (width, height))

    # Resize only for display
    display_original = resize_for_display(image)
    display_resized = resize_for_display(resized)

    cv2.imshow("Original Image", display_original)
    cv2.imshow("Resized Image", display_resized)

    cv2.waitKey(0)
    cv2.destroyAllWindows()


# ================= ROTATE =================

elif choice == "2":

    angle = float(input("Enter rotation angle: "))

    height, width = image.shape[:2]

    center = (width // 2, height // 2)

    matrix = cv2.getRotationMatrix2D(
        center,
        angle,
        1.0
    )

    rotated = cv2.warpAffine(
        image,
        matrix,
        (width, height)
    )

    # Resize only for display
    display_original = resize_for_display(image)
    display_rotated = resize_for_display(rotated)

    cv2.imshow("Original Image", display_original)
    cv2.imshow("Rotated Image", display_rotated)

    cv2.waitKey(0)
    cv2.destroyAllWindows()


# ================= BLUR =================

elif choice == "3":

    kernel = int(input("Enter kernel size (odd number): "))

    if kernel % 2 == 0:
        print("Kernel size must be odd!")
        exit()

    blurred = cv2.GaussianBlur(
        image,
        (kernel, kernel),
        0
    )

    # Resize only for display
    display_original = resize_for_display(image)
    display_blurred = resize_for_display(blurred)

    cv2.imshow("Original Image", display_original)
    cv2.imshow("Blurred Image", display_blurred)

    cv2.waitKey(0)
    cv2.destroyAllWindows()


else:
    print("Invalid choice!")
```
