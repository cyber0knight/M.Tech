```python
import cv2

# Read image
image = cv2.imread("Man.jpg")

if image is None:
    print("Image not found!")
    exit()

# Convert to grayscale
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Callback function
def nothing(x):
    pass

window = "Canny Edge Detection"

# Create window
cv2.namedWindow(window, cv2.WINDOW_NORMAL)

# Create trackbars
cv2.createTrackbar("Threshold 1", window, 100, 500, nothing)
cv2.createTrackbar("Threshold 2", window, 200, 500, nothing)

while True:

    # Get threshold values
    threshold1 = cv2.getTrackbarPos("Threshold 1", window)
    threshold2 = cv2.getTrackbarPos("Threshold 2", window)

    # Apply Canny
    edges = cv2.Canny(gray, threshold1, threshold2)

    # Show result
    cv2.imshow(window, edges)

    # ESC key to exit
    if cv2.waitKey(1) == 27:
        break

cv2.destroyAllWindows()
```
